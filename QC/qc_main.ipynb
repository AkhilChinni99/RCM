```python
def main():
    # Read widgets
    qc_ids_str = dbutils.widgets.get("qc_ids")   # e.g. "1,2,3"
    job_name   = dbutils.widgets.get("job_name")
    run_id     = dbutils.widgets.get("run_id")

    # Convert QC IDs to list
    qc_id_list = [int(qc.strip()) for qc in qc_ids_str.split(",") if qc.strip()]

    print("Job Name :", job_name)
    print("Run ID   :", run_id)
    print("QC IDs   :", qc_id_list)

    # Execute QC for each qc_id
    for qc_id in qc_id_list:
        print(f"Running QC ID: {qc_id}")
        run_qc(qc_id, job_name, run_id)


if __name__ == "__main__":
    main()


#In Workflow
#qc_ids=1,2,3
#job_name={{job.name}}
#run_id={{job.run_id}}
```
