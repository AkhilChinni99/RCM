```python
try:
    volume_path = dbutils.widgets.get("volume_path")
    landing_table = dbutils.widgets.get("landing_table")
except Exception as e:
    print(f"Error initializing variables: {e}")
    raise
```

```python

try:
    spark.sql(f"""
        COPY INTO {landing_table}
        FROM (
            SELECT
                CAST(_id AS STRING) AS _id,
                CAST(claim_type AS STRING) AS claim_type,
                CAST(patient_control_no AS STRING) AS patient_control_no,
                CAST(patient_ctl_no AS STRING) AS patient_ctl_no,
                CAST(claim_charge AS STRING) AS claim_charge,
                CAST(patient_name AS STRING) AS patient_name,
                CAST(create_date AS TIMESTAMP) AS create_date,
                CAST(payer_name AS STRING) AS payer_name,
                CAST(payer_sequence AS STRING) AS payer_sequence,
                CAST(patient_dob AS TIMESTAMP) AS patient_dob,
                CAST(statement_start AS TIMESTAMP) AS statement_start,
                CAST(from_date AS TIMESTAMP) AS from_date,
                CAST(statement_end AS TIMESTAMP) AS statement_end,
                CAST(through_date AS TIMESTAMP) AS through_date,
                CAST(id AS BIGINT) AS id,
                CAST(patient_id AS STRING) AS patient_id,
                CAST(trans_type AS STRING) AS trans_type,
                CAST(_metadata.file_name AS STRING) AS _file_name,
                CAST(current_timestamp() AS TIMESTAMP) AS _load_timestamp
            FROM '{volume_path}'
        )
        FILEFORMAT = JSON
        FORMAT_OPTIONS ('multiLine' = 'true')
        COPY_OPTIONS ('mergeSchema' = 'true')
    """)
except Exception as e:
    print(f"Error loading data into {landing_table}: {e}")
    raise
```
