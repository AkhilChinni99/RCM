```python
start_dt=dbutils.widgets.get("start_dt")
end_dt=dbutils.widgets.get("end_dt")
```

```python
from datetime import datetime, timedelta
start_dt = (datetime.strptime(start_dt, '%Y-%m-%d') - timedelta(days=1)).strftime('%Y-%m-%d')
end_dt = (datetime.strptime(end_dt, '%Y-%m-%d') - timedelta(days=1)).strftime('%Y-%m-%d')
```

```python
dbutils.jobs.taskValues.set(key="start_dt", value=start_dt)
dbutils.jobs.taskValues.set(key="end_dt", value=end_dt)
```
