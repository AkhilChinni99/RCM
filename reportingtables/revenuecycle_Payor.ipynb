```python
revenuecycle_payor_path=dbutils.widgets.get("revenuecycle_payor_path")
payor_path=dbutils.widgets.get("payor_path")
cubeserviceofficetxnpayorcategory_path=dbutils.widgets.get("cubeserviceofficetxnpayorcategory_path")
```

```python
spark.sql(
    f"""
    TRUNCATE TABLE {revenuecycle_payor_path};
    """
)
```

```python
spark.sql(
    f"""
INSERT INTO {revenuecycle_payor_path} (
    payor_key,
    payor_id,
    payor_name,
    payor_address1,
    payor_address2,
    payor_city,
    payor_state,
    payor_zip,
    payor_program,
    payor_type_code,
    payor_type_name,
    payor_type_description,
    payor_category
)
SELECT
    p.PayorKey AS payor_key,
    p.PayorID AS payor_id,
    p.`Name` AS payor_name,
    p.Address1 AS payor_address1,
    p.Address2 AS payor_address2,
    p.City AS payor_city,
    p.`State` AS payor_state,
    p.Zip AS payor_zip,
    p.PayorProgram AS payor_program,
    c.PayorCategoryCode AS payor_type_code,
    c.PayorCategoryName AS payor_type_name,
    c.PayorCategoryDescription AS payor_type_description,
    c.PayorCategoryCategory AS payor_category
FROM {payor_path} p
LEFT JOIN {cubeserviceofficetxnpayorcategory_path} c 
    ON p.TypeID = c.PayorCategoryCode
WHERE 
    (
        c.PayorCategoryKey NOT IN (46, 16, 21, 13, 24, 3, 30, 29)
        OR c.PayorCategoryDescription IS NULL
    )
"""
)
```
