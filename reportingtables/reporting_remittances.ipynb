```python
reporting_remittances_path=dbutils.widgets.get("reporting_remittances_path")
remittances_path=dbutils.widgets.get("remittances_path")
date_path=dbutils.widgets.get("date_path")
```

```python
spark.sql(
    f"""
    TRUNCATE TABLE {reporting_remittances_path};
    """
)
```

```python
spark.sql(
    f"""
INSERT INTO {reporting_remittances_path} (
    id,
    reporting_week_ending_date_key,
    create_date_key,
    payment_date_key,
    statement_start_date_key,
    statement_end_date_key,
    source_system_key,
    office_key,
    payor_key,
    client_key,
    enterprise_category_key,
    denial_type_key,
    payer_icn,
    ability_payor_name,
    patient_name,
    member_id,
    adjustment_codes,
    remark_code,
    patient_ctl_no,
    invoice_number,
    charge,
    total_service_adjustments,
    claim_payment,
    underpayment_amount,
    check_eft_number,
    landing_update_datetime,
    landing_update_by
)
SELECT 
    ID AS id,
    Reporting_Week_Ending_Date_Key AS reporting_week_ending_date_key,
    Create_Date_Key AS create_date_key,
    Payment_Date_Key AS payment_date_key,
    Statement_Start_Date_Key AS statement_start_date_key,
    Statement_End_Date_Key AS statement_end_date_key,
    Source_System_Key AS source_system_key,
    Office_Key AS office_key,
    Payor_Key AS payor_key,
    Client_Key AS client_key,
    Enterprise_Category_Key AS enterprise_category_key,
    Denial_Type_Key AS denial_type_key,
    Payer_ICN AS payer_icn,
    Ability_Payor_Name AS ability_payor_name,
    Patient_Name AS patient_name,
    Member_ID AS member_id,
    Adjustment_Codes AS adjustment_codes,
    Remark_Code AS remark_code,
    PatientCtlNo AS patient_ctl_no,
    Invoice_Number AS invoice_number,
    Charge AS charge,
    Total_Service_Adjustments AS total_service_adjustments,
    Claim_Payment AS claim_payment,
    Underpayment_Amount AS underpayment_amount,
    Check_EFTNumber AS check_eft_number,
    landing_update_datetime,
    landing_update_by
FROM {remittances_path}
WHERE reporting_week_ending_date_key <=
CASE
    WHEN dayofweek(current_date()) <= 4
    THEN (
        SELECT date_format(date_add(WeekEndingDate, -14), 'yyyyMMdd')
        FROM {date_path}
        WHERE CalendarDate = current_date()
    )
    ELSE (
        SELECT date_format(date_add(WeekEndingDate, -7), 'yyyyMMdd')
        FROM {date_path}
        WHERE CalendarDate = current_date()
    )
END
"""
)
```
