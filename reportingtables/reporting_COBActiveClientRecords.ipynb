```python
spark.sql(
    f""" 
INSERT INTO {reporting_COBActiveClientRecords} (
    client_id,
    client_full_name,
    client_status,
    last_visit_date,
    office_number,
    reimbursement_office_number,
    secondary_payor_id,
    secondary_payor_name,
    secondary_payor_program,
    referral_id,
    subscriber_id,
    payor_name_other_bill_to,
    other_bill_to,
    payor_order,
    payor_effective_date,
    payor_term_date,
    cob_status,
    cob_type,
    cob_recd_date,
    cob_eff_date,
    cob_term_date,
    cob_date_needed,
    cob_level,
    cob_denial_reason,
    split_decision,
    limited_benefit,
    benefit_date,
    home_plan,
    split_decision_details,
    is_mco_referral,
    mco_coordinator,
    note_text,
    note_title,
    notes_author_full_name,
    notes_date_entered,
    comments,
    payor_status,
    auth_id,
    auth_number,
    auth_date_entered,
    comment_1,
    comment_2,
    auth_last_update,
    auth_updated_by,
    date_entered,
    last_update,
    updated_by,
    bill_hold_status,
    bill_hold_reason,
    level,
    bill_hold_date,
    office_name,
    office_division,
    ar_balance,
    client_cob_id
)
SELECT
    C.Client_Id AS client_id,
    concat(CD.FirstName, ' ', CD.LastName) AS client_full_name,
    CD.SystemStatusCode AS client_status,
    C.Last_Visit_Date AS last_visit_date,
    C.Office_Number AS office_number,
    C.Reimbursement_Office_Number AS reimbursement_office_number,
    C.Payor_Id AS secondary_payor_id,
    C.Payor_Name AS secondary_payor_name,
    C.Payor_Program AS secondary_payor_program,
    C.Referral_Id AS referral_id,
    C.Subscriber_Id AS subscriber_id,
    C.Payor_Name__Other_Bill_To_ AS payor_name_other_bill_to,
    C.Other_Bill_To AS other_bill_to,
    C.Payor_Order AS payor_order,
    C.Payor_Effective_Date AS payor_effective_date,
    C.Payor_Term_Date AS payor_term_date,
    C.COB_Status AS cob_status,
    C.COB_Type AS cob_type,
    C.COB_Recd_Date AS cob_recd_date,
    C.COB_Eff_Date AS cob_eff_date,
    C.COBTerm_Date AS cob_term_date,
    C.COB_Date_Needed AS cob_date_needed,
    C.COB_Level AS cob_level,
    C.COB_Denial_Reason AS cob_denial_reason,
    C.Split_Decision AS split_decision,
    C.Limited_Benefit AS limited_benefit,
    C.Benefit_Date AS benefit_date,
    C.Home_Plan AS home_plan,
    C.Split_Decision_Details AS split_decision_details,
    C.Is_MCO_Referral AS is_mco_referral,
    C.MCO_Coordinator AS mco_coordinator,
    C.Note_Text AS note_text,
    C.Note_Title AS note_title,
    C.Employee_Full_Name AS notes_author_full_name,
    C.Notes_Date_Entered AS notes_date_entered,
    C.Comments AS comments,
    C.Payor_Status AS payor_status,
    C.Auth_Id AS auth_id,
    C.Auth_Number AS auth_number,
    C.Auth_Date_Entered AS auth_date_entered,
    C.Comment_1 AS comment_1,
    C.Comment_2 AS comment_2,
    C.Auth_Last_Update AS auth_last_update,
    C.Auth_Updated_By AS auth_updated_by,
    C.Date_Entered AS date_entered,
    C.Last_Update AS last_update,
    C.Updated_By AS updated_by,
    C.Bill_Hold_Status AS bill_hold_status,
    C.Bill_Hold_Reason AS bill_hold_reason,
    C.Level AS level,
    C.Bill_Hold_Date AS bill_hold_date,
    C.Office_Name AS office_name,
    C.Office_Division AS office_division,
    C.AR_Balance AS ar_balance,
    C.Client_Cob_Id AS client_cob_id
FROM {coordinationofbenefits_path} C
INNER JOIN {client_path} CD
    ON C.Client_Key = CD.ClientKey
   AND Deleted = 0
WHERE
(
    year(to_date(C.COB_Date_Needed, 'MM-dd-yyyy')) >= year(current_date())
    OR
    (
        year(to_date(C.COB_Date_Needed, 'MM-dd-yyyy')) = year(current_date()) - 1
        AND C.COB_Status IN ('RCVDPART', 'PENDING')
    )
)
AND C.COB_Date_Needed IS NOT NULL;
"""
)
```
