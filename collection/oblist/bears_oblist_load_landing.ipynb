```python
file_list = dbutils.widgets.get("file_list")
oblist_path = dbutils.widgets.get("oblist_path")
landing_table = dbutils.widgets.get("landing_table")
```

```python
import ast
file_list = ast.literal_eval(file_list)
for file_name in file_list:
    file_path = oblist_path + file_name
    df_oblist = spark.read.option("header", "true").option("delimiter", "\t").csv(file_path)
    df_oblist.createOrReplaceTempView("oblist_view")

    spark.sql(f"""
    INSERT INTO {landing_table}
    SELECT
    `OFFICE` AS office,
    `OABB` AS oabb,
    `CLIENTNO` AS clientno,
    `CLIENTNAME` AS clientname,
    `CLIENTFIRSTNAME` AS clientfirstname,
    `INVNO` AS invno,
    `INVDATE` AS invdate,
    `PAYORNAME` AS payorname,
    `Net Revenue` AS net_revenue,
    `ORIGBILL` AS origbill,
    `GROUP#` AS group_num,
    `billing team3` AS billing_team3,
    `TEAM` AS team,
    `payor type code` AS payor_type_code,
    `PAYORTYPE` AS payortype,
    `BILLTO` AS billto,
    `BILL TO PROGRAM` AS bill_to_program,
    `TOTALDUE` AS totaldue,
    `Current` AS current,
    `4 - 7 WKS` AS four_seven_wks,
    `8 - 13 WKS` AS eight_thirteen_wks,
    `14 wks to Reserve` AS fourteen_wks_to_reserve,
    `Up for Reserve` AS up_for_reserve,
    `Prior Reserve` AS prior_reserve,
    `NET` AS net,
    `Specialty` AS specialty,
    `PAYOR CATEGORY` AS payor_category,
    `CLIENT STATE` AS client_state,
    `COLLECTOR NAME` AS collector_name,
    `last post date` AS last_post_date,
    `client status` AS client_status,
    `company` AS company,
    `last payment date` AS last_payment_date,
    `Serv W/E Date` AS serv_w_e_date,
    `net payments` AS net_payments,
    `net adjustments` AS net_adjustments,
    `1st visit date` AS first_visit_date,
    `last visit date` AS last_visit_date,
    `total days serviced` AS total_days_serviced,
    `npi number` AS npi_number,
    current_timestamp() AS _load_timestamp, 
    "{file_name[:-4]}" AS _file_name
    FROM oblist_view
    """)    
```
