```python
run_type = dbutils.widgets.get("run_type")
trigger_date = dbutils.widgets.get("trigger_date")
start_date = dbutils.widgets.get("start_date")
end_date = dbutils.widgets.get("end_date")
```

```python
from datetime import datetime, timedelta
```

```python
if run_type == "daily-run":
    start_date = (datetime.strptime(trigger_date, '%Y-%m-%d') - timedelta(days=2)).strftime('%Y-%m-%d')
    end_date = trigger_date
    fetch_date = (datetime.strptime(trigger_date, '%Y-%m-%d') - timedelta(days=1)).strftime('%Y-%m-%d')
    dbutils.jobs.taskValues.set(key="start_date", value=start_date)
    dbutils.jobs.taskValues.set(key="end_date", value=end_date)
    dbutils.jobs.taskValues.set(key="fetch_date", value=fetch_date)
elif run_type == "re-load":
    dbutils.jobs.taskValues.set(key="start_date", value=start_date)
    dbutils.jobs.taskValues.set(key="end_date", value=end_date)
elif run_type == "full-load":
    start_date = "1990-01-01"
    end_date = trigger_date
    dbutils.jobs.taskValues.set(key="start_date", value=start_date)
    dbutils.jobs.taskValues.set(key="end_date", value=end_date)
elif run_type == "job-triggered":
    dbutils.jobs.taskValues.set(key="start_date", value=start_date)
    dbutils.jobs.taskValues.set(key="end_date", value=end_date)
    fetch_date = (datetime.strptime(trigger_date, '%Y-%m-%d') - timedelta(days=1)).strftime('%Y-%m-%d')
    dbutils.jobs.taskValues.set(key="fetch_date", value=fetch_date)
else:
    raise ValueError(f"Invalid run_type: {run_type}")
```
